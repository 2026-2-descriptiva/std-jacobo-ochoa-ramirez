```python
import csv
import gzip
import sqlite3
from pathlib import Path

from pytest import approx

ACTIVITY_DIR = "LAB_09_sql_con_sqlite3"

CODE_DIR = ACTIVITY_DIR + "/src"

TABLES = {
    "tbl0": "K0 CHAR(1), c01 INT, c02 INT, c03 CHAR(4), c04 FLOAT",
    "tbl1": "K0 CHAR(1), K1 INT, c12 FLOAT, c13 INT, c14 DATE, c15 FLOAT, c16 CHAR(4)",
    "tbl2": "K1 INT, c21 FLOAT, c22 INT, c23 DATE, c24 FLOAT, c25 CHAR(5)",
}


def normalize(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return value
from pathlib import Path


ACTIVITY_DIR = Path("LAB_09_sql_con_sqlite3")
CODE_DIR = ACTIVITY_DIR / "src"


def query(number):
    connection = sqlite3.connect(":memory:")

    try:
        for table, columns in TABLES.items():
            connection.execute(f"CREATE TABLE {table} ({columns})")

            path = ACTIVITY_DIR / "data" / f"{table}.csv.gz"

            with gzip.open(
                path,
                "rt",
                encoding="utf-8-sig",
                newline="",
            ) as file:
                rows = list(csv.reader(file))

            placeholders = ", ".join("?" for _ in columns.split(","))
            connection.executemany(
                f"INSERT INTO {table} VALUES ({placeholders})",
                rows,
            )

        sql_path = CODE_DIR / f"pregunta_{int(number):02d}.sql"
        sql = sql_path.read_text(encoding="utf-8")

        rows = connection.execute(sql).fetchall()
        return [
            tuple(normalize(value) for value in row)
            for row in rows
        ]
    finally:
        connection.close()

query("01")
```
